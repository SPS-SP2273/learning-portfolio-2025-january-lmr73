# Task 4 - Data Cleaning

1. Shortening Names for the Universities
2. Replacing "na" with np.nan

```python
df['university'].unique()         # Checking the unique university names present
```

```python
replace_info = {'Nanyang Technological University': 'NTU',
                'National University of Singapore': 'NUS',
                'Singapore Management University': 'SMU',
                'Singapore Institute of Technology': 'SIT',
                'Singapore University of Technology and Design': 'SUTD',
                'Singapore University of Social Sciences': 'SUSS',           # Dictionary matching the full University names to abbreviated form, to change names
                'na': np.nan,
               'np.nan': np.nan}               # Dictionary matching "na" and "np.nan" to np.nan, to change str to np.nan           

#using dictionary above, replace full University names with abbreviation, replace "na" and "np.nan" to np.nan
df.replace(to_replace=replace_info, inplace=True)
```

```python
(df=='na').sum().sum()
```

```python
df.iloc[0:25,0:5]    # Checking if changes are made
```

3. Changing basic_monthly_median type to float.

```python
df.info()        # Checking what type it is
```

```python
# Changing column Dtype from object to float
new_type_info = {'basic_monthly_mean': 'float'}
df = df.astype(new_type_info)
df = df.astype({'degree':'str'})
df.info()      # Checking if changes are made
```

### Purpose of shortening Uni name:
- For ease of readability
- Reduce human error in the code

### Purpose of replacing "na" with np.nan:
- Cleaning the df.
- Changing "na" string into np.nan (which is not a number, and is recognised as null value), ensures that the system recognises that it is a missing value, so that .agg() can run - a str would return an error, but nan is acceptable, and skipped in .agg().
- This is because pandas operations will skip null values by default.

## Purpose of changing column type from object into float:
- We need it to be float to calculate the mean, median as needed in Task 5 (mean and median operations cannot be done correctly on objects as they cannot be interpreted as numeric values).
- Float used instead of Int so that precision is kept for values with decimal places.

# Task 5 - Analyse the Data

```python
# Summary statistics based on the values in basic_monthly_median
df['basic_monthly_median'].describe(include='all')
```

```python
# Table of data that shows the minimum, maximum, mean for the basic_monthly_median
x = df.groupby(by=["year", "university"])[['basic_monthly_median']].agg(["count", "min", "max", "mean", "std"]).round(3)

x
```

```python
y = df.groupby(by=['degree','university'])[['basic_monthly_mean']].agg(['std'])
y
```

```python
for degree in df['degree']:
    if degree.startswith('Accountancy'):
        df.replace(to_replace = {degree:'Accountancy'}, inplace=True)
    else: continue 
```

### Comments on code:
- .groupby() allows us to organise the data in this specified manner, by year, then by University. 
- .agg() is used because we want specific outcomes like count, min, max, mean, std.
- Using .describe() gives us the values above & 25,50,75 percentile, which is not wanted.
- We wanted to customise the output, so we used .agg().
- agg() is for calculating a lot of summary statistics in one line of code rather than np.mean(), np.median()

### Why is this step needed?
- To allow us to obtain information for basic_monthly_median specifically, grouped in this particular manner, so that we can plot the bar graph based on these values, in this grouping.

# Task 6 - Time to plot

```python
import matplotlib.pyplot as plt
```

```python
df_pivot = df.pivot_table(values='basic_monthly_median', index='year', columns='university', aggfunc='mean')
df_pivot.plot(kind='bar')
plt.ylabel('Median Salary')
plt.xlabel('year')
plt.xticks(rotation=0)
plt.legend(loc=(1.05,0.4))
plt.tight_layout()
plt.show()
```

### Using the column information to plot bar chart
- Allows us to visualise the trend of median salaries of different Universities over time.
- aggfunc function is set as "mean" because we only want to plot the mean value.
- Mean values were presented as they account for all values in the dataset, reflecting the overall distribution. This makes it more useful when analysing trends, compared to the median, which only considers the middle value
